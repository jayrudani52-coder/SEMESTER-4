```python
import requests
from bs4 import BeautifulSoup
url="https://www.politifact.com/factchecks"
result=requests.get(url)
print(result.status_code)
content=result.text
soup=BeautifulSoup(content,'html.parser')
print(soup.prettify())
```

```python
with open("politifact.txt","w",encoding="utf-8") as file:
    file.write(content)
```

```python
statement=[]
for i in movie:
    a=i.find("div",class_="m-statement__quote").text.strip()
    print(a)
    statement.append(a)
print(statement)
```

```python
source=[]
for i in movie:
    s=i.find("div",class_="m-statement__meta").find("a").text.strip()
    source.append(s)
print(source)
```

```python
date=[]
for i in movie:
    d=i.find("footer").text.strip()
    date.append(d)
print(date)
```

```python
date=[]
for i in movie:
    d=i.find("footer").text.split("• ")[-1].strip()
    date.append(d)
print(date)
```

```python
links=[]
root="https://politifact.com"
for i in movie:
    l=i.find("div",class_="m-statement__quote").find("a")["href"].strip()
    links.append(root+l)
print(links)
```

```python
import pandas as pd
df=pd.DataFrame({"Statement":statement,"Source":source,"Date":date,"Link":links})
df
```

```python
upperframe=[]
for page in range(1,5):
    url="https://www.politifact.com/factchecks/?page="+str(page)
    result=requests.get(url)
    content=result.text
    soup=BeautifulSoup(content,'html.parser')
    movie=soup.find_all("li",class_="o-listicle__item")
    frame=[]
    for i in movie:
        statement=i.find("div",class_="m-statement__quote").text.strip()
        root="https://www.politifact.com"
        root+=i.find("div",class_="m-statement__quote").find("a")["href"].strip()
        date=i.find("footer").text.split("• ")[-1].strip()
        source=i.find("div",class_="m-statement__meta").find("a").text.strip()
        frame.append((statement,root,date,source))
    upperframe.extend(frame)
print(upperframe)
```

```python
import pandas as pd
df=pd.DataFrame(upperframe,columns=["Statement","Links","Date","Source"])
df
```
