```python
import requests
from bs4 import BeautifulSoup

base_url = "https://subslikescript.com"
url = f"{base_url}/movies"

# Step 1: Get the listing page
result = requests.get(url)
soup = BeautifulSoup(result.text, 'html.parser')

# Step 2: Find all movie links
box = soup.find('article', class_='main-article')
links = [a['href'] for a in box.find_all('a', href=True)]

# Step 3: Visit each script link and save
for link in links:
    script_url = base_url + link
    result = requests.get(script_url)
    script_soup = BeautifulSoup(result.text, 'html.parser')

    title_tag = script_soup.find('h1')
    script_box = script_soup.find('div', class_='full-script')

    if title_tag and script_box:
        title = title_tag.get_text(strip=True)
        script_text = script_box.get_text("\n", strip=True)

        # Clean filename
        filename = title.replace(":", "").replace("/", "-") + ".txt"

        # Save script to file
        with open(filename, "w", encoding="utf-8") as file:
            file.write(script_text)

        print(f"Saved: {filename}")
    else:
        print(f"Skipped: {script_url}")
```
