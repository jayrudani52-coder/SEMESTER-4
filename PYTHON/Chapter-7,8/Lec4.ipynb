```python
import requests,json
def current_weather():
    api_key="75b80d48c71af37eb15d22e0d0250469"
    city=input("Enter the City name: ")
    request_url=f"http://api.openweathermap.org/geo/1.0/direct?q={city}&appid={api_key}"
    response=requests.get(request_url)
    coordinates=response.json()
    lat=coordinates[0]["lat"]
    lon=coordinates[0]["lon"]
    request_url1=f"https://api.openweathermap.org/data/2.5/weather?lat={lat}&lon={lon}&appid={api_key}"
    response1=requests.get(request_url1)
    weather_data=response1.json();
    print(json.dumps(weather_data,indent=5))
    des=weather_data["weather"][0]["description"]
    temp=weather_data["main"]["temp"]
    press=weather_data["main"]["pressure"]
    hum=weather_data["main"]["humidity"]
    vis=weather_data["visibility"]
    win_sp=weather_data["wind"]["speed"]
    print("====================")
    print("Current Weather data of current city :",city);
    print("--------------------")
    print("Weather Description :",des)
    print("Temperature :",temp)
    print("Pressure :",press)
    print("Humidity :",hum)
    print("Visibility :",vis)
    print("Wind Speed :",win_sp)
current_weather()
```

```python
import requests
def five_day_forecast():
    api_key="75b80d48c71af37eb15d22e0d0250469"
    city=input("Enter the City name: ")
    request_url=f"http://api.openweathermap.org/geo/1.0/direct?q={city}&appid={api_key}"
    response=requests.get(request_url)
    coordinates=response.json()
    lat=coordinates[0]["lat"]
    lon=coordinates[0]["lon"]
    request_url1=f"https://api.openweathermap.org/data/2.5/forecast?lat={lat}&lon={lon}&appid={api_key}"
    response1=requests.get(request_url1)
    five_day=response1.json();
    return five_day
five_day=five_day_forecast()
```

```python
import pandas as pd
data={"Date_Time":[],"Description":[],"Temp":[],"Pressure":[],"Humidity":[],"Visibility":[],"Wind Speed":[]}
for i in five_day["list"]:
    data["Date_Time"].append(i["dt_txt"])
    data["Description"].append(i["weather"][0]["description"])
    data["Temp"].append(i["main"]["temp"])
    data["Pressure"].append(i["main"]["pressure"])
    data["Humidity"].append(i["main"]["humidity"])
    data["Visibility"].append(i["visibility"]) 
    data["Wind Speed"].append(i["wind"]["speed"])
df=pd.DataFrame(data)
df
```
